# Fonction Radar chart

```python
from openpyxl import load_workbook
```

```python
wb=load_workbook("../output/test/test.xlsx")
```

```python
ws_tdb=wb["TDB"]
ws_sct=wb["SCATTER"]
```

```python
from openpyxl.chart import (
    RadarChart,
    Reference,
)

c1=RadarChart()
x_values=Reference(ws_sct, min_col=2,min_row=1,max_col=4, max_row=1)
y_values=Reference(ws_sct,min_col=2,max_col=4, min_row=2, max_row=5)
c1.add_data(y_values, titles_from_data=True)
c1.set_categories(x_values)
c1.y_axis.delete=True

ws_sct.add_chart(c1, "H1")

wb.save("../output/test/feat-4.xlsx")
```

## Test de normalisation avec la méthode du min max

```python
from openpyxl_boardgames.utils import get_data, join_df, choose_columns, clean_data, save_data
from openpyxl_boardgames.config import URL1, URL2, URL3, COLUMNS, OBJECT_FLOAT, FILL_NA, NEW_TYPE
```

```python
df_main = get_data(URL1, sep=';')
df_sec = get_data(URL2, sep=',')
df_thr = get_data(URL3, sep=',')
print(f"Données récupérées")

df_main=join_df(df_left=df_main, df_right=df_sec,key_left='ID',key_right='game_id',how_join='left')
df_main=join_df(df_left=df_main, df_right=df_thr,key_left='ID',key_right='id',how_join='left')
print("Jointure effectuée")

df_main=choose_columns(df_main, COLUMNS)

df_main=clean_data(df_main, OBJECT_FLOAT, FILL_NA, NEW_TYPE)
```

```python
df_norm=df_main["Rating Average"]
df_norm
```
